## Loading in the data

```python
import sklearn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

The code below imports the pickle module which is the file type that my data uses. It opens one file and reads it to a variable called data. Then, I displayed data to see what the format looked like.

```python
import pickle

# Open the pickle file in read-binary mode
with open('../data/keypoints2d/gBR_sBM_cAll_d04_mBR0_ch01.pkl', 'rb') as file:
    data = pickle.load(file)

# Now, 'data' contains the object stored in the .pkl file
keypoints = data['keypoints2d'][0]
timestamps = data['timestamps']

print(keypoints)
```

In this cell, I plotted the x and y-coordinates over time for the data I loaded in.

```python

import matplotlib.pyplot as plt

# Assuming keypoints is of shape (num_frames, num_keypoints*2), 
# where num_keypoints*2 represents x and y coordinates for each keypoint.
body_parts = ["nose","left_eye","right_eye","left_ear","right_ear","left_shoulder","right_shoulder","left_elbow",
             "right_elbow","left_wrist","right_wrist","left_hip","right_hip","left_knee","right_knee","left_ankle",
             "right_ankle"]

num_keypoints = keypoints.shape[1] // 2  # Number of keypoints

# Plotting x-coordinates of each keypoint over time
plt.figure(figsize=(12, 7))
for i in range(num_keypoints):
    plt.plot(timestamps, keypoints[:, 2*i], label=body_parts[i])

plt.xlabel('Time (s)')
plt.ylabel('X-Coordinate')
plt.title('X-Coordinates of Keypoints Over Time')
plt.legend()
plt.show()

# Plotting y-coordinates of each keypoint over time
plt.figure(figsize=(12, 7))
for i in range(num_keypoints):
    plt.plot(timestamps, keypoints[:, 2*i+1], label=body_parts[i])

plt.xlabel('Time (s)')
plt.ylabel('Y-Coordinate')
plt.title('Y-Coordinates of Keypoints Over Time')
plt.legend()
plt.show()
```
